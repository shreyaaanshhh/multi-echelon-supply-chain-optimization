# Data Cleaning & Preparation

**Goal:** Clean and merge raw order data with carrier freight rates to build a reliable, cost-verified dataset for the optimization model in `02_lp_optimization.ipynb`.

**Source:** [Kaggle - Supply Chain Logistics Problem](https://www.kaggle.com/datasets/anisseezzebdi/supply-chain-logistics-problem)

## 1. Load the raw data

```python
import pandas as pd
xls = pd.ExcelFile("../data/raw/Supply chain logistics problem.xlsx", engine="openpyxl")
print(xls.sheet_names)
```

```python
wh_costs = pd.read_excel(xls, sheet_name = "WhCosts")
print(wh_costs.head())

wh_capacities = pd.read_excel(xls, sheet_name = "WhCapacities")
print(wh_capacities.head())

freight_rates = pd.read_excel(xls, sheet_name = "FreightRates")
print(freight_rates.head())

plant_ports = pd.read_excel(xls, sheet_name = "PlantPorts")
print(plant_ports.head())
```

## 2. Explore the order data

Checking the structure of `OrderList` before cleaning - this is where we discover that the dataset has only **one destination port**, meaning `Customer` (not `Port`) is the real demand-side node in the network. This directly shaped the decision to model a **Plant → Customer** network rather than a deeper multi-port chain.

```python
order_list = pd.read_excel(xls, sheet_name = "OrderList")
print(order_list.head())
print(order_list.shape)

demand_by_port = order_list.groupby("Destination Port")["Unit quantity"].sum().reset_index()
print(demand_by_port)
```

```python
print(order_list["Customer"].nunique())
demand_by_customer = order_list.groupby("Customer")["Unit quantity"].sum().reset_index()
print(demand_by_customer.head(10))
```

## 3. Merge orders with freight rates

`OrderList` has no direct shipping cost - `FreightRates` has cost, but indexed by Carrier + Origin/Destination Port + weight bracket, not by order. Before merging, we check carrier coverage between the two tables.

```python
matching_carriers = order_list["Carrier"].isin(["V444_0", "V444_1"])
print(matching_carriers.sum())
```

```python
print(order_list["Carrier"].value_counts())
```

**Finding:** carrier `V44_3` (854 orders, ~9.3%) has no matching entries at all in `FreightRates`. This is a genuine gap in the source data, not a processing error - these orders are excluded below since no valid freight rate exists for them.

## 4. Merge, filter by weight bracket, and deduplicate

- Merge on Carrier + Origin Port + Destination Port (inner join - only keep orders with a real rate match).
- Filter to the row where the order's actual weight falls inside the rate's weight bracket.
- Some orders still match multiple overlapping brackets — keep the cheapest valid rate per order.

```python
merged = order_list.merge(freight_rates, left_on = ["Carrier", "Origin Port", "Destination Port"], right_on = ["Carrier", "orig_port_cd", "dest_port_cd"], how = "inner")
print(merged.shape)

final = merged[(merged["Weight"] >= merged["minm_wgh_qty"]) & (merged["Weight"] <= merged["max_wgh_qty"])]
print(final.shape)

final_sorted = final.sort_values("rate")
final_dedup =  final_sorted.drop_duplicates(subset = "Order ID", keep = "first")
print(final_dedup.shape)
print(final.shape)
```

**Result:** `final_dedup` contains **6,991 of 9,215 raw orders (~75.9%)**, each with a single, verified, cheapest-available freight rate. This is the clean dataset used for all downstream cost, demand, and optimization work.

## 5. Export cleaned data for the optimization notebook

```python
final_dedup.to_csv("../data/processed/cleaned_orders.csv", index=False)
print("Saved:", final_dedup.shape)
```
