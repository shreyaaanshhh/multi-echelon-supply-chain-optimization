```python
#initialize the problem
prob = LpProblem("Problem_!", LpMaximize)

#decision param
x = LpVariable.dicts("x", [1,2], lowBound = 0, cat = "Continuous")

#objective function
prob += 10 * x[1] + 20 * x[2]

#constraints
prob += -x[1] + 2*x[2] <= 15
prob += 5*x[1] + 3*x[2] <= 45
prob += x[1] + x[2] <= 12

#solve the problem
status = prob.solve()
print(LpStatus[status])

#display the results
for var in prob.variables():
    print(f" {var.name} = {var.varValue}")
```

```python
groves = ["Mt. Dora", "Eustis", "Clermont"]
plants = ["Ocala", "Orlando", "Leesburg"]
supply = {"Mt. Dora": 275000, "Eustis": 400000, "Clermont": 300000}
capacity = {"Ocala": 200000, "Orlando": 600000, "Leesburg": 225000}

cost = {
    "Mt. Dora": {"Ocala": 21, "Orlando": 50, "Leesburg": 40},
    "Eustis":   {"Ocala": 35, "Orlando": 30, "Leesburg": 22},
    "Clermont": {"Ocala": 55, "Orlando": 20, "Leesburg": 25}
}

prob = LpProblem("Tansportation", LpMinimize)

x = LpVariable.dicts("ship", (groves, plants), lowBound = 0, cat = "Continuous")

prob += lpSum(cost[g][p] * x[g][p] for g in groves for p in plants)

for g in groves:
    prob += lpSum(x[g][p] for p in plants) == supply[g]

for p in plants:
    prob += lpSum(x[g][p] for g in groves)<=capacity[p]

status = prob.solve()
print(LpStatus[status])
for g in groves:
    for p in plants:
        if x[g][p].varValue > 0:
            print(f"ship from {g} to {p} = {x[g][p].varValue}")
print("Total bushels shipped = ", value(prob.objective))

```
